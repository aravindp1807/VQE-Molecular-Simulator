## ⚛️ Step 4: Run VQE Quantum Simulation & Energy Curve Sweep
Simulates ground state electronic energy using Qiskit VQE with IBM Quantum API Key pre-configured.

```python
from vqe_colab_simulator import VQESimulator, MoleculeVisualizer
import matplotlib.pyplot as plt

DEFAULT_IBM_TOKEN = "7hk4AmU8jgQqqDjkkY4fpYD-3h6VtCjSPJqfSU0xmf9f"
simulator = VQESimulator(ibm_token=DEFAULT_IBM_TOKEN)

h2_atoms = [
    {"element": "H", "x": 0.0, "y": 0.0, "z": 0.0},
    {"element": "H", "x": 0.74, "y": 0.0, "z": 0.0}
]

# Sweep interatomic distance from 0.4 Å to 2.2 Å
distances, exact_energies, vqe_energies = simulator.simulate_energy_curve(
    atoms=h2_atoms,
    bond_range=(0.4, 2.2),
    num_points=10,
    use_ibm=False  # Set True to send jobs to IBM Quantum Cloud
)

# Plot Potential Energy Surface (PES) Dissociation Curve
fig_pes = MoleculeVisualizer.plot_potential_energy_curve(distances, exact_energies, vqe_energies, molecule_name="H2")
plt.show()
```
